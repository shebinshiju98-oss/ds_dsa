```python
class Stack:
    def __init__(self):
        self.data = []
    def push(self, item):
        self.data.append(item)
    def display(self):
        if self.is_empty():
            print("Stack is empty")
        else:
            for i in range(len(self.data) - 1, -1, -1):
                print(self.data[i], end=" ")
            print()
    def pop(self):
        if self.is_empty():
            print("Stack is empty")
            return None
        return self.data.pop()
    def is_empty(self):
        return len(self.data) == 0
    def peek(self):
        if not self.is_empty():
            return self.data[-1]
        else:
            print("Stack is empty")

s = Stack()
s.push(10)
s.push(20)
s.push(30)
print("Stack elements (top to bottom):")
s.display()
print("\nPopping elements:")
print(s.pop())
print(s.pop())
print(s.pop())
s.pop()
print("\nStack after popping all elements:")
s.display()
```

```python
# Create a class called order.4 ppties order id,date,price and qty.

class Order:
    def __init__(self, order_id, date, price, qty):
        self.order_id = order_id
        self.date = date
        self.price = price
        self.qty = qty
    def display(self):
        print("Order ID:", self.order_id)
        print("Date:", self.date)
        print("Price:", self.price)
        print("Quantity:", self.qty)
        print("-------------------")

o1 = Order(101, "15-07-2026", 500, 2)
o2 = Order(102, "16-07-2026", 750, 1)
o3 = Order(103, "17-07-2026", 1200, 3)
o4 = Order(104, "18-07-2026", 300, 5)
o5 = Order(105, "19-07-2026", 950, 2)
o1.display()
o2.display()
o3.display()
o4.display()
o5.display()
```

```python
from collections import deque
order_queue=deque()
order_queue.append(o1)
order_queue.append(o2)
order_queue.append(o3)
order_queue.append(o4)
order_queue.append(o5)

# for order in order_queue:
#     order.display()

```

```python
while order_queue:
    cur=order_queue.popleft()
    print("Processing order")
    cur.display()
```

```python
class Patient:
    def __init__(self,id,name,dept,type):
        self.id = id
        self.name = name
        self.dept = dept
        self.type = type
    def display(self):
        print("Patient ID:", self.id)
        print("Name:", self.name)
        print("Department:", self.dept)
        print("Type:", self.type)
        print("-------------------")
    
```

```python
p1 = Patient(101, "Arun Kumar", "Cardiology", "Emergency")
p2 = Patient(102, "Meera Nair", "Orthopedics", "Regular")
p3 = Patient(103, "Rahul Das", "Neurology", "Emergency")
p4 = Patient(104, "Anjali Menon", "Dermatology", "Regular")
p5 = Patient(105, "Vivek Sharma", "ENT", "Emergency")

patients_info = [p1, p2, p3, p4, p5]

for patient in patients_info:
    patient.display()
```

```python
import heapq
patients=[] 
heapq.heappush(patients,(3,p1))
heapq.heappush(patients,(2,p2))
heapq.heappush(patients,(1,p3))
heapq.heappush(patients,(4,p4))
heapq.heappush(patients,(5,p5))

while patients:
    pr, p = heapq.heappop(patients)
    print("Processing Priority:", pr)
    p.display()
```
