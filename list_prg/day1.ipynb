```python
#target=5 o/p=[2,3] target=14 op=[5,9] target=6 op=[1,5]
num = [2, 3, 4, 1, 5, 9]
target = int(input())
for i in range(len(num)):
    for j in range(i + 1, len(num)):
        if num[i] + num[j] == target:
            print([num[i], num[j]])
            break
    else:
        continue
    break
```
